```python
from attogpt.config import Config
from attogpt.utils import clean_dataset
from attogpt.tokenizer import BPETokenizer
```

```python
data = clean_dataset("../data/raw_corpus.txt","../data/clean_corpus.txt")
print(len(data), "characters")
```

```python
config = Config()
tokenizer = BPETokenizer(config)
ids = tokenizer.tokenize(data)
print(len(ids), "tokens")
```

```python
tokenizer.to_file(ids, "../data/vocabulary.json", "../data/corpus_ids.npy")
```
