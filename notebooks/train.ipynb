```python
import numpy as np
from attogpt.config import Config
from attogpt.model import AttoGPT
from attogpt.train import Train

config = Config(seq_len=64, d_model=108)
rng = np.random.default_rng(config.seed)
model = AttoGPT(config, rng)

corpus = np.load("../data/plato/corpus_ids.npy")

trainer = Train(
    model, config, corpus, rng,
    batch_size=32,
    lr=1e-3,
    steps=5000,
)

# trainer = Train.from_checkpoint("../data/plato/checkpts/checkpt.pkl", corpus, batch_size=64, lr=3e-4, steps=4000)

trainer.train(save_path="../data/plato/checkpts/checkpt.pkl", eval_freq=100, patience=20)
```
